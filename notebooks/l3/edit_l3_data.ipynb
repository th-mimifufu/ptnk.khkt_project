```python
import polars as pl

# infer_schema_length=None scans the WHOLE file to find the correct number of columns
df = pl.read_excel("../../data/l3-transcript-data.xlsx")
print(df.head())
```

```python
df = df.with_columns(
    pl.when(pl.col("major_code").str.contains("V"))
    .then(pl.lit("Đại trà"))
    .when(pl.col("major_code").str.contains("BP"))
    .then(pl.lit("Đào tạo tại Bình Phước"))
    .when(pl.col("major_code").str.contains("A"))
    .then(pl.lit("Đào tạo bằng tiếng Anh"))
    .when(pl.col("major_code").str.contains("N"))
    .then(pl.lit("Đào tạo bằng tiếng Nhật"))
    .when(pl.col("major_code").str.contains("DD"))
    .then(pl.lit("Thuộc ngành Công nghệ thực phẩm (đào tạo bằng tiếng Việt)"))
    .otherwise(pl.lit(None))
    .alias("study_program")
)
```

```python
df = df.with_columns(
    pl.lit("Học Bạ").alias("admission_type"),
    pl.lit("Sử dụng phương thức xét tuyển học bạ").alias("admission_type_name")
)
```

```python
qsb = pl.read_csv("../../data/priority-admission-data.csv")
```

```python
qsb = qsb.filter(pl.col("uni_code")=="QSB")
```

```python
qsb = qsb.with_columns(
    # 1. Thay thế chữ "Ưu Tiên" thành "Tổng Hợp" trong cột admission_code
    pl.col("admission_code").str.replace("Ưu Tiên", "Tổng Hợp"),

    # 2. Set lại toàn bộ cột admission_type_name thành chuỗi cố định
    pl.lit("Sử dụng phương thức tổng hợp").alias("admission_type_name")
).drop("subject_combination")
```

```python
qsb
```

```python
qsb = qsb.with_columns(
    pl.col("admission_code")
    .str.split("-")          # 1. Tách chuỗi dựa trên dấu "-"
    .list.get(1)             # 2. Lấy phần tử thứ 2 (index 1) trong list kết quả
    .alias("admission_code") # 3. Ghi đè lại vào cột cũ
)
```

```python
qsb
```

```python
import polars as pl

# Giả sử bạn đã có 2 dataframe là df và qsb

# 1. Thực hiện Merge (Left Join)
# - left_on="major_code": Key của df
# - right_on="admission_code": Key của qsb
# - suffix="_qsb": Đổi tên các cột trùng từ qsb để tránh lỗi duplicate column
merged = df.join(
    qsb, 
    left_on="major_code", 
    right_on="admission_code", 
    how="left", 
    suffix="_qsb"
)

# 2. Xác định các cột chung cần điền dữ liệu (Fill Null)
# Lấy danh sách cột có ở cả 2 bảng (trừ cột key major_code/admission_code ra vì tên khác nhau)
common_cols = [col for col in df.columns if col in qsb.columns]

# 3. Thực hiện Fill Null & Select
# Logic: Với mỗi cột chung, update giá trị bằng cột _qsb nếu cột gốc bị null
final_df = merged.with_columns([
    pl.col(c).fill_null(pl.col(f"{c}_qsb")) for c in common_cols
]).select(
    # Giữ lại toàn bộ cột ban đầu của df
    df.columns
)

# Hiển thị kết quả
print(final_df)
```

```python
final_df = final_df.with_columns(
    # Cập nhật cột admission_type
    pl.when(pl.col("uni_code") == "QSB")
    .then(pl.lit("Tổng Hợp"))
    .otherwise(pl.col("admission_type"))
    .alias("admission_type"),

    # Cập nhật cột admission_type_name
    pl.when(pl.col("uni_code") == "QSB")
    .then(pl.lit("Sử dụng phương thức xét tuyển tổng hợp"))
    .otherwise(pl.col("admission_type_name"))
    .alias("admission_type_name")
)
```

```python
final_df = final_df.with_columns(
    pl.concat_str(
        [
            pl.col("uni_code"), 
            pl.col("major_code"), 
            pl.col("admission_type")
        ],
        separator="-"
    ).alias("admission_code")
)
```

```python
final_df = final_df.with_columns(
    pl.col("uni_type").cast(pl.Int64)
)
```

```python
import polars as pl
import uuid
from datetime import datetime

# 1. Cấu hình kết nối
connection_uri = "postgresql://myuser:secret@localhost:5433/node_app_dev"

# 2. Hàm sinh UUID và thêm Timestamp
# Vì Polars xử lý vector, cách nhanh nhất để tạo UUID unique cho từng dòng là dùng List Comprehension
uuid_list = [uuid.uuid4() for _ in range(final_df.height)]
current_time = datetime.now().isoformat()
df_to_write = final_df.with_columns(
    # Thêm cột UUID
    pl.Series(name="id", values=uuid_list),
    
    # Thêm cột thời gian
    pl.lit(current_time).alias("created_at"),
    pl.lit(current_time).alias("updated_at")
)


# 3. Ghi vào DB (Schema: machine_learning)
try:
    df_to_write.write_database(
        table_name="machine_learning.l3_transcript_new", # Đã sửa lỗi schema ở đây
        connection=connection_uri,
        if_table_exists="replace",
        engine="sqlalchemy"
    )
    print("✅ Đã tạo UUID và load dữ liệu thành công!")
except Exception as e:
    print(f"❌ Có lỗi xảy ra: {e}")
```

```python
from sqlalchemy import create_engine, text
engine = create_engine(connection_uri)
with engine.connect() as conn:
    # Lệnh SQL biến cột id thành KHÓA CHÍNH (Primary Key)
    conn.execute(text("ALTER TABLE machine_learning.l3_transcript_new ADD PRIMARY KEY (id);"))
    conn.commit()
    print("✅ Đã set Primary Key thành công!")
```
